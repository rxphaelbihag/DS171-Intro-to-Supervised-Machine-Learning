## **A. Load the Data (Digits Dataset)**

The digits dataset is one of datasets scikit-learn comes with that do not require the downloading of any file from some external website. The code below will load the digits dataset.

```python
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
# Print to show there are 1797 images (8 by 8 images for a dimensionality of 64)
print("Image Data Shape" , digits.data.shape)
# Print to show there are 1797 labels (integers from 0–9)
print("Label Data Shape", digits.target.shape)
```

**Plot the training datasets**

Import numpy and matplotlib.pyplot. Set a figure size of $(20, 4)$, iterate through the first 5 image samples, reshape each 64-element array into an $8 \times 8$ matrix, and plot them in grayscale with their target labels as titles:

```python
# Your source code here

import numpy
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 4))

# iterate over the first 5 image samples
for i in range(5):
    matrix = digits.data[i].reshape(8, 8)       # reshapes the vector
    plt.matshow(matrix, cmap='gray')            # grayscale plot
    plt.title(f"Image of {digits.target[i]}")

plt.show()
```

**Split the train dataset.**


```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.25, random_state=0)
```

**Scikit-learn 4-Step Modeling Pattern (Digits Dataset)**

**Step 1. Import the model you want to use**

In sklearn, all machine learning models are implemented as Python classes.
You may use the Logistic Regression model.

```python
# Your source code here
from sklearn.linear_model import LogisticRegression
```

**Step 2. Make an instance of the Model.**

Create an instance of LogisticRegression named logisticRegr with default parameters:

```python
# Your source code here
model = LogisticRegression()
```

**Step 3. Training the model on the data, storing the information learned from the data**

*   Model is learning the relationship between digits (x_train) and labels (y_train).
*   Fit your model using the training features (x_train) and labels (y_train).
*   Reshape the first test image (x_test[0]) into a 2D array using .reshape(1, -1) and predict its class using logisticRegr.predict().
*   Pass the first 10 test samples (x_test[0:10]) into logisticRegr.predict().
*   Predict labels for all test samples in x_test and store them in a variable named predictions.
*   Compute the model's accuracy on the test set using .score(), store it in score, and print it.





```python
# Your source code here
# fitting the model to the training sets
model.fit(x_train, y_train)

# predicting the first test image
pred_1 = model.predict(x_test[0].reshape(1, -1))
print("Prediction first test:", pred_1)

# predicting the first 10 test samples
pred_10 = model.predict(x_test[0:10])
print("Prediction on first 10 tests:", pred_10)

# predicting all test samples
predictions = model.predict(x_test)
print("Prediction on all tests:", predictions)

# computing the model's accuracy
score = model.score(x_test, y_test)
print("Model's Score:", score)
```

**Step 4. Creating a Confusion Matrix**

Create the confusion matrix of your predictions. Include labels for both actual and predicted values. Use the libraries below.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import metrics
```

- Compute the confusion matrix using metrics.confusion_matrix(y_test, predictions), store it in a variable, and print it.
- Render a Seaborn heatmap of the confusion matrix with numerical annotations (fmt=".3f"), square cells, axis labels, and a title displaying the test accuracy score.
- Plot an alternative visualization of your confusion matrix using plt.imshow() with the 'Pastel1' color map, custom ticks (0–9), labeled axes, and annotated cell counts using nested loops.

```python
# Your source code here

# confusion matrix from sklearn.metrics
conf_matrix = metrics.confusion_matrix(y_test, predictions)
print("Confusion Matrix:\n", conf_matrix)

# plot confusion matrix using seaborn's heatmap
sns.heatmap(conf_matrix, annot=True, fmt=".3f")
plt.title(f"Confusion Matrix Test Acc: {score:.3f}")
plt.xlabel("Predicted Labels")
plt.ylabel("True Labels")
plt.show()
```

Output the data type of the mnist variable. Check the data types of data, categories, feature_names, and target attributes. Then check the shapes of the mnist.data and mnist.target.

```python
# Your source code here

print("mnist type:", type(mnist))
print("="*10)
print("data type:", type(mnist.data))
print("categories type:", type(mnist.categories))
print("feature_names type:", type(mnist.feature_names))
print("target type:", type(mnist.target))
print("="*10)
print("mnist.data shape:", mnist.data.shape)
print("mnist.target shape:", mnist.target.shape)
```

In this part, check first the target element.

- Display the first target element.
- Convert it to integer by casting mnist.target array values to integers using .astype('int').
- Split mnist.data and integer-casted targets (mnist.target.astype('int')) into X2_train, X2_test, y2_train, and y2_test.
- Display the shapes of X2_train and X2_test.

```python
# Your source code here

first_target = mnist.target.iloc[0] if hasattr(mnist.target, 'iloc') else mnist.target[0]
print("First target element:", first_target)
print("Type before converting:", type(first_target))

# convert to integer
mnist.target = mnist.target.astype('int')

# split the dataset
X2_train, X2_test, y2_train, y2_test = train_test_split(
    mnist.data, mnist.target, test_size=0.2, random_state=42
)

# display the shapes
print("\nX2_train shape:", X2_train.shape)
print("X2_test shape:", X2_test.shape)
```

### Plot Class Distribution Histograms

Plot side-by-side histograms using plt.hist() to compare class frequency distributions between training and test sets.

```python
# Your source code here
plt.hist(
    [y2_train, y2_test],
    label=["Training Set", "Test Set"],
    rwidth=0.8,
)

plt.title("Class Frequency Distribution: Training vs. Test Set")
plt.xlabel("Digit Class")
plt.ylabel("Frequency")
plt.xticks(range(10))
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

Instantiate a LogisticRegression classifier clf2 with Lasso regularization (penalty='l1'), saga solver, C=50, parallel workers (n_jobs=5), verbosity enabled, and tolerance 0.01:

```python
# Your source code here
clf2 = LogisticRegression(penalty='l1', solver='saga', C=50.0,
                          n_jobs=5, verbose=1, tol=0.01)
```

Use the %%time cell magic command to time the model training process on X2_train and y2_train.

```python
# Your source code here
%%time
clf2.fit(X2_train, y2_train)
```

Inspect the .classes_ attribute of clf2. Then, check the shape of clf2.coef_. Print rounded weight values for class index 1. Finally, output the model intercept values and print the number of iterations before reaching the tolerance.

```python
# Your source code here
print("Model Classes:", clf2.classes_)
print("Shape of Coef:", clf2.coef_.shape)
print(clf2.coef_[1].round(decimals=2))
print("="*20)
print("Model Intercept", clf2.intercept_)
print("Model Iterations:", clf2.n_iter_)
```

**28 x 28 Spatial Heatmaps**

- Copy clf2.coef_, reshape each class weight vector into a $28 \times 28$ image matrix, and display them in a $2 \times 5$ subplot grid using the RdBu divergent color map.
- Print model predictions alongside ground truth labels for the first 9 test samples.
- Calculate and display model training score.
- Calculate, store in score2, and display testing score.

```python
# Your source code here
fig, axes = plt.subplots(2, 5, figsize=(20, 8))

# define the numbers
rows = 2
cols = 5
counter = 0

# loops over the numbers
for row in range(rows):
    for col in range(cols):
        matrix = clf2.coef_[counter].reshape(28, 28)    # reshapes into a matrix
        axes[row, col].matshow(matrix, cmap='RdBu')     # plots the matrix
        axes[row, col].set_title(f"Class {counter}")    # sets the title
        counter+=1

plt.tight_layout()
plt.show()
```

```python
pred_9 = clf2.predict(X2_test[:9])
print("Predictions:", pred_9)
print("Actual True:", list(y2_test[:9]))
print("="*20)
score = clf2.score(X2_train, y2_train)
print(f"Train Score: {score:.4f}")
print("="*20)
score2 = clf2.score(X2_test, y2_test)
print(f"Test Score: {score2:.4f}")
```

**MNIST Confusion Matrix**

- Predict labels on X2_test, store them as predictions2, and calculate the target confusion matrix cm.
- Plot a heatmap of the confusion matrix using Seaborn with numeric values formatted to 4 significant digits (fmt='0.4g').

```python
# Your source code here
predictions2 = clf2.predict(X2_test)
cm = metrics.confusion_matrix(y2_test, predictions2)

# plot confusion matrix using seaborn's heatmap
sns.heatmap(cm, annot=True, fmt="0.4g")
plt.title(f"Confusion Matrix Test Acc: {score:.4f}")
plt.xlabel("Predicted Labels")
plt.ylabel("True Labels")
plt.show()
```

###Analysis:

- Q1. In plotting spatial heatmaps for each digit class, you reshaped the coefficient vector back to a $28 \times 28$ grid. How does increasing the feature dimensionality from $64$ to $784$ impact both the visual interpretability of these weight matrices and the computational complexity during optimization?

> Shifting from $64$ to $784$ dimensions allows for a much better visual interpretability as it increases the resolution of the image. We can see that it went from pixelated ($8 \times 8$) to better quality ($28 \times 28$). Furthermore, increasing the dimensions also increases the computational complexity during optimization as there will be more features to be calculated; more data to work with, more flops, therefore more complex.

- Q2. In your configuration of L1 penalty in LogisticRegression, a regularization strength paramater of C=50, and a tolerance of tol=0.01 were applied. Since standard Logistic Regression is a binary classifier, explain how scikit-learn handles multi-class prediction (digits 0–9) behind the scenes.

> Logistic Regression is a standard binary classifier. However, Sklearn's `LogisticRegression` class automatically handles multiclass regression. As stated in Sklearn's documentation, "All classifiers in scikit-learn do multiclass classification out-of-the-box. You don’t need to use the sklearn.multiclass module unless you want to experiment with different multiclass strategies." With that, the documentation also specified that they optimize multiclass problems on multinomial loss.

- Q3. Analyze the effect of L1 regularization on the weight vector: what would happen to the coefficient values and model performance if you significantly decreased C (e.g., from C=50 to C=0.01), and why might L1 regularization be particularly useful when dealing with high-dimensional image pixels?

> The parameter `C` is the inverse of regularization strength, which means that smaller values specify stronger regularization. By having the C value significantly decreased, we configure the L1 regularization to be drastic. This leads to the model being stricter to its criteria of "zero-ing" out elements of the weight vector. Especially for high-dimensional data such as high-res images, L1 regularization is useful as it can help reduce computational complexity by focusing only on the highly useful features and reduce the chance of overfitting by minimizing the degrees of freedom the model will "memorize" features.
